```r
rds_path <- "../backend/uploads/job_1769501813220_ccec5e89-75ed-4859-82ce-32f524b08143/healthy_breast_preprocessed_RNA_peaks_only_7_8.rds"
assay_name <- "RNA"
h5ad_path  <- "../backend/uploads/job_1769501813220_ccec5e89-75ed-4859-82ce-32f524b08143/healthy_breast_preprocessed_RNA_peaks_only_7_8.h5ad"
```

```r
data_seurat <- readRDS(rds_path)
h5seurat_file <- sub("\\.rds$", ".h5Seurat", rds_path)
if (h5ad_path=="") {
    h5ad_file <- sub("\\.h5Seurat$", paste0("_", assay_name, ".h5ad"), h5seurat_file)
} else {
    h5ad_file <- h5ad_path
}
data_seurat[[assay_name]] <- as(object = data_seurat[[assay_name]], Class = "Assay")
SeuratDisk::SaveH5Seurat(
    object    = data_seurat,
    filename  = h5seurat_file,
    overwrite = TRUE
  )
SeuratDisk::Convert(
    h5seurat_file,
    dest      = "h5ad",
    assay     = assay_name,
    filename  = h5ad_file,
    overwrite = TRUE
  )
```

```r
# Check versions of Seurat and SeuratObject
pkgs <- c("Seurat", "SeuratObject", "SeuratDisk")
versions <- sapply(pkgs, function(p) {
  if (requireNamespace(p, quietly = TRUE)) {
    as.character(utils::packageVersion(p))
  } else {
    NA
  }
})
for (p in pkgs) {
  v <- versions[p]
  if (is.na(v)) {
    cat(p, ": not installed\n")
  } else {
    cat(p, ": ", v, "\n", sep = "")
  }
}
```
