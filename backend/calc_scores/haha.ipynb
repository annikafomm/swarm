```python
import scanpy as sc
import pandas as pd
from scipy import io
import os

dir = os.path.join("..", "datasets_prepro")

files_per_dir = []
for file in os.listdir(dir):
    if file.endswith('.h5ad') and file.startswith("G"):
        files_per_dir.append(os.path.join(dir, file))

adatas = [sc.read_h5ad(file) for file in files_per_dir]
for i in range(len(adatas)):
    print(files_per_dir[i])
    print(adatas[i].n_vars, adatas[i].n_obs)
    print(len(set(adatas[i].obs['leiden'])))
```

```python
i=0
adata = adatas[i]

new_filename = files_per_dir[i].replace('.h5ad', '')
# Write matrix
io.mmwrite(f"{new_filename}_X.mtx", adata.X)
# Save row (cell) names
pd.Series(adata.obs_names).to_csv(f"{new_filename}_cells.txt", index=False, header=False)

adata.var.to_csv(f"{new_filename}_var.csv")
```
