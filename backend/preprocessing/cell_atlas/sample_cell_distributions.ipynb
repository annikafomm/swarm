```python
import pandas as pd
import matplotlib.pyplot as plt

# keep only needed columns
obs = Global_raw.obs[Global_raw.obs["combinedID"]!="na"][["region", "cell_type", "combinedID"]].copy()

# optional: drop missing values
obs = obs.dropna(subset=["region", "cell_type", "combinedID"])

# make a label for each region-combinedID combination
obs["region_combinedID"] = (
    obs["region"].astype(str) + " | " + obs["combinedID"].astype(str)
)
counts = (
    obs.groupby(["region", "combinedID"])
    .size()
    .reset_index(name="n_cells")
)

counts["region_combinedID"] = (
    counts["region"].astype(str) + " | " + counts["combinedID"].astype(str)
)

# sort by count descending
counts = counts.sort_values("n_cells", ascending=False)

plt.figure(figsize=(max(12, len(counts) * 0.25), 5))
plt.bar(counts["region_combinedID"], counts["n_cells"])
plt.xlabel("region | combinedID")
plt.ylabel("Number of cells")
plt.title("Cells per region-combinedID combination")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

```

```python

stacked = (
    obs.groupby(["region_combinedID", "cell_type"])
    .size()
    .unstack(fill_value=0)
)

# sort bars by total cell count
stacked = stacked.loc[stacked.sum(axis=1).sort_values(ascending=False).index]

ax = stacked.plot(
    kind="bar",
    stacked=True,
    figsize=(max(12, len(stacked) * 0.25), 6),
    width=0.9
)

ax.set_xlabel("region | combinedID")
ax.set_ylabel("Number of cells")
ax.set_title("Cells per region-combinedID combination, stacked by cell_type")
plt.xticks(rotation=90)
plt.legend(title="cell_type", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
