```python
import scanpy as sc
import squidpy as sq

adata = sc.read_h5ad("../outputs/prepro_defaults/GSM6592049_M2_prepro_defaults.h5ad")
```

```python
sq.pl.spatial_scatter(
    adata,
    library_id="spatial",
    shape=None,
    color=[
        "leiden",
    ],
    wspace=0.4,
)
```

```python
sq.pl.centrality_scores(adata, cluster_key="leiden", figsize=(16, 5))
```

```python
for i in range(9):
    sq.pl.co_occurrence(
        adata,
        clusters=f"{i}",
        cluster_key="leiden",
        figsize=(10, 10),
    )
```

```python
sq.pl.nhood_enrichment(
    adata,
    cluster_key="leiden",
    figsize=(8, 8),
    title="Neighborhood enrichment adata"
)
```
