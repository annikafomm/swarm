```python
def row_normalize(df, eps=1e-12):
    x = df.clip(lower=0).astype(float)
    s = x.sum(axis=1)
    return x.div(s + eps, axis=0)

c2l_comp = row_normalize(c2l)   # should stay basically the same
tg_comp  = row_normalize(tg)    # important step
```

```python
from scipy.stats import pearsonr, spearmanr
from scipy.spatial.distance import jensenshannon

def cosine_sim(x, y, eps=1e-12):
    return np.dot(x, y) / ((np.linalg.norm(x) * np.linalg.norm(y)) + eps)

def topk_jaccard(x, y, k=3):
    a = set(np.argsort(x)[-k:])
    b = set(np.argsort(y)[-k:])
    return len(a & b) / len(a | b)

def compare_per_spot(df1, df2):
    rows = []
    for spot in df1.index:
        x = df1.loc[spot].values.astype(float)
        y = df2.loc[spot].values.astype(float)

        rows.append({
            "spot_id": spot,
            "pearson": pearsonr(x, y)[0] if np.std(x) > 0 and np.std(y) > 0 else np.nan,
            "spearman": spearmanr(x, y)[0],
            "cosine": cosine_sim(x, y),
            "js_distance": jensenshannon(x + 1e-12, y + 1e-12),
            "top1_same": int(df1.loc[spot].idxmax() == df2.loc[spot].idxmax()),
            "top3_jaccard": topk_jaccard(x, y, k=3),
        })
    return pd.DataFrame(rows).set_index("spot_id")

spot_metrics = compare_per_spot(c2l_comp, tg_comp)
spot_metrics.describe()
```

```python
def compare_per_celltype(df1, df2):
    rows = []
    for ct in df1.columns:
        x = df1[ct].values.astype(float)
        y = df2[ct].values.astype(float)

        rows.append({
            "cell_type": ct,
            "pearson": pearsonr(x, y)[0] if np.std(x) > 0 and np.std(y) > 0 else np.nan,
            "spearman": spearmanr(x, y)[0],
            "mean_abs_diff": np.mean(np.abs(x - y)),
            "rmse": np.sqrt(np.mean((x - y) ** 2)),
        })
    return pd.DataFrame(rows).set_index("cell_type").sort_values("spearman", ascending=False)

ct_metrics = compare_per_celltype(c2l_comp, tg_comp)
ct_metrics
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_true = c2l_comp.values.ravel()
y_pred = tg_comp.values.ravel()

overall_metrics = {
    "MAE": mean_absolute_error(y_true, y_pred),
    "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
    "R2": r2_score(y_true, y_pred),
    "Pearson_all": np.corrcoef(y_true, y_pred)[0, 1],
}

overall_metrics
```

```python
def lin_ccc(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mx, my = x.mean(), y.mean()
    vx, vy = x.var(), y.var()
    cov = np.mean((x - mx) * (y - my))

    return (2 * cov) / (vx + vy + (mx - my) ** 2 + 1e-12)

rows = []
for ct in c2l_comp.columns:
    x = c2l_comp[ct].values
    y = tg_comp[ct].values

    rows.append({
        "cell_type": ct,
        "MAE": mean_absolute_error(x, y),
        "RMSE": np.sqrt(mean_squared_error(x, y)),
        "R2": r2_score(x, y),
        "Pearson": pearsonr(x, y)[0] if np.std(x) > 0 and np.std(y) > 0 else np.nan,
        "Spearman": spearmanr(x, y)[0],
        "CCC": lin_ccc(x, y),
    })

eval_ct = pd.DataFrame(rows).set_index("cell_type").sort_values("CCC", ascending=False)
eval_ct
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score

thr = 0.05

rows = []
for ct in c2l_comp.columns:
    y_true_bin = (c2l_comp[ct] >= thr).astype(int)
    y_score = tg_comp[ct].values

    if y_true_bin.nunique() < 2:
        auc = np.nan
        ap = np.nan
    else:
        auc = roc_auc_score(y_true_bin, y_score)
        ap = average_precision_score(y_true_bin, y_score)

    rows.append({
        "cell_type": ct,
        "AUROC": auc,
        "AUPRC": ap,
        "prevalence": y_true_bin.mean(),
    })

presence_eval = pd.DataFrame(rows).set_index("cell_type").sort_values("AUPRC", ascending=False)
presence_eval
```
