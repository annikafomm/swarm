```python
import pandas as pd
import requests
from io import StringIO
```

```python

go0030199_genes = [
    "CRTAP","P3H4","EMILIN1","ADAMTS7","ANTXR2","COL1A1","COL1A2","COL2A1",
    "COL3A1","COL4A1","COL4A2","COL4A3","COL4A4","COL4A5","COL4A6","COL5A1",
    "COL5A2","COL6A1","COL7A1","COL11A1","COL11A2","COL12A1","COMP","ADAMTS14",
    "CHADL","CYP1B1","AEBP1","ADAMTS19","DPT","EXT1","FOXC1","FOXC2","COLGALT2",
    "FMOD","OPTC","VPS33B","GREM1","P4HA3","ERO1A","EFEMP2","ANXA2","COL28A1",
    "LOX","LOXL1","LOXL2","LUM","MIR29B1","MMP11","NF1","DDR2","P4HA1","COL5A3",
    "SERPINF2","PLOD1","PLOD2","ATP7A","SELENON","RB1","FKBP10","VIPAS39","P3H1",
    "SFRP2","SCX","BMP1","TGFB2","TGFBR1","TLL1","TLL2","TNXB","COL14A1","PXDN",
    "COLGALT1","ADAMTS12","LOXL4","LOXL3","SERPINH1","PLOD3","ADAMTS3","ADAMTS2"
]

url = "https://omnipathdb.org/interactions/"
params = {
    "genesymbols": 1,
    "datasets": "collectri",
    "organisms": 9606,
    "targets": ",".join(go0030199_genes),
    "fields": "sources,references"
}

r = requests.get(url, params=params, timeout=60)
r.raise_for_status()

raw = pd.read_csv(StringIO(r.text), sep="\t")

gene_tf_df = (
    raw[["target_genesymbol", "source_genesymbol"]]
    .rename(columns={"target_genesymbol": "gene", "source_genesymbol": "TF"})
    .dropna()
    .drop_duplicates()
    .sort_values(["gene", "TF"])
    .reset_index(drop=True)
)

gene_tf_df
```

```python
gene_tf_df.to_csv("objects_AVN_fibroblast/gene_tf_relations_AVN_fibroblasts.csv", index=False)
```
